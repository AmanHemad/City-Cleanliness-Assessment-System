```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# Load model
model = YOLO("/home/venom/Documents/CSE/CSE_y3s2/Mini_Project/City-Cleanliness-Assessment-System/City_cleanliness_dataset/python_scripts/best.pt")

# Run prediction
results = model.predict(
    source="/home/venom/Documents/CSE/CSE_y3s2/Mini_Project/City-Cleanliness-Assessment-System/City_cleanliness_dataset/python_scripts/clean_dataset/000002.jpg",
    conf=0.25
)

# Process result (single image → results[0])
r = results[0]

# Read original image
img = cv2.imread(r.path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Loop through detections
for box in r.boxes:
    x1, y1, x2, y2 = map(int, box.xyxy[0])
    conf = float(box.conf[0])
    cls = int(box.cls[0])

    # Print coordinates
    print(f"Class: {cls}, Confidence: {conf:.2f}")
    print(f"Bounding Box: ({x1}, {y1}), ({x2}, {y2})\n")

    # Draw rectangle
    cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)

    # Label text
    label = f"{cls} {conf:.2f}"
    cv2.putText(img, label, (x1, y1 - 5),
                cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

# Show image
plt.imshow(img)
plt.axis("off")
plt.show()
```
