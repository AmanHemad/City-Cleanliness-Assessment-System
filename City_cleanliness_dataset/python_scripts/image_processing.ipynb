```python
import cv2
import numpy as np
import os
from tqdm import tqdm
from lama_cleaner.model_manager import ModelManager
from lama_cleaner.schema import Config

INPUT_FOLDER = "clean_dataset"
OUTPUT_FOLDER = "output_images_cleanest"

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# load AI model
model = ModelManager(name="lama", device="cpu")

config = Config(
    ldm_steps=20,
    hd_strategy="Resize",
)

def detect_diagonal_watermark_mask(image):

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # detect edges
    edges = cv2.Canny(gray, 40, 120)

    # detect lines
    lines = cv2.HoughLinesP(edges, 1, np.pi/180, 120,
                            minLineLength=80,
                            maxLineGap=20)

    mask = np.zeros(gray.shape, dtype=np.uint8)

    if lines is not None:
        for line in lines:

            x1,y1,x2,y2 = line[0]

            angle = np.degrees(np.arctan2(y2-y1, x2-x1))

            # detect diagonal lines
            if 30 < abs(angle) < 70:

                cv2.line(mask,(x1,y1),(x2,y2),255,5)

    kernel = np.ones((5,5),np.uint8)
    mask = cv2.dilate(mask,kernel,iterations=3)

    return mask


def enhance_image(img):

    # edge preserving smoothing
    smooth = cv2.edgePreservingFilter(img, flags=1,
                                      sigma_s=60,
                                      sigma_r=0.4)

    # detail enhance
    detail = cv2.detailEnhance(smooth,
                               sigma_s=10,
                               sigma_r=0.15)

    # sharpen
    kernel = np.array([[0,-1,0],
                       [-1,5,-1],
                       [0,-1,0]])

    sharpen = cv2.filter2D(detail,-1,kernel)

    return sharpen


for file in tqdm(os.listdir(INPUT_FOLDER)):

    path = os.path.join(INPUT_FOLDER,file)

    img = cv2.imread(path)

    if img is None:
        continue

    # detect watermark
    mask = detect_diagonal_watermark_mask(img)

    # AI inpainting
    cleaned = model(img, mask, config)

    # restore quality
    restored = enhance_image(cleaned)

    output_path = os.path.join(OUTPUT_FOLDER,file)

    cv2.imwrite(output_path, restored)

print("Processing complete.")
```

```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from IPython.display import display
from ipywidgets import Button, HBox, VBox, Output, Text
from concurrent.futures import ThreadPoolExecutor

# -------- CHANGE THIS --------
base_folder = "/home/venom/Documents/CSE/CSE_y3s2/Mini_Project/City-Cleanliness-Assessment-System/City_cleanliness_dataset/cleanliness_dataset"
review_log = "reviewed_images.txt"
# -----------------------------

image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

# Thread pool for background loading
executor = ThreadPoolExecutor(max_workers=2)

# Cache for prefetched images
cache = {}

# Load already reviewed images
reviewed = set()
if os.path.exists(review_log):
    with open(review_log, "r") as f:
        reviewed = set(line.strip() for line in f)

# Collect images
subfolders = [
    os.path.join(base_folder, f)
    for f in os.listdir(base_folder)
    if os.path.isdir(os.path.join(base_folder, f))
]

image_paths = []

for folder in subfolders:
    for file in os.listdir(folder):
        if file.lower().endswith(image_extensions):
            path = os.path.join(folder, file)
            if path not in reviewed:
                image_paths.append(path)

print("Images remaining:", len(image_paths))

index = 0
stopped = False
out = Output()


# -------- PREFETCH FUNCTION --------
def load_image(path):
    return mpimg.imread(path)


def prefetch(idx):
    if idx < len(image_paths):
        path = image_paths[idx]
        if path not in cache:
            future = executor.submit(load_image, path)
            cache[path] = future


def get_image(path):
    if path in cache:
        img = cache[path].result()
        del cache[path]
        return img
    return mpimg.imread(path)
# ----------------------------------


def log_review(path):
    with open(review_log, "a") as f:
        f.write(path + "\n")


def show_image():
    global index, stopped

    out.clear_output()

    if stopped:
        with out:
            print("Review stopped.")
        return

    if index >= len(image_paths):
        with out:
            print("All images reviewed.")
        return

    img_path = image_paths[index]
    folder_name = os.path.basename(os.path.dirname(img_path))

    # Prefetch NEXT image
    prefetch(index + 1)

    with out:
        print(f"Image {index+1} / {len(image_paths)}")
        print("Folder:", folder_name)
        print("Path:", img_path)

        img = get_image(img_path)

        plt.figure(figsize=(6,6))
        plt.imshow(img)
        plt.axis("off")
        plt.show()


def keep_image(b=None):
    global index
    path = image_paths[index]

    log_review(path)
    index += 1
    show_image()


def delete_image(b=None):
    global index
    path = image_paths[index]

    try:
        os.remove(path)
        print("Deleted:", path)
    except:
        print("Delete failed:", path)

    log_review(path)
    index += 1
    show_image()


def skip_folder(b=None):
    global index
    current_folder = os.path.dirname(image_paths[index])

    while index < len(image_paths) and os.path.dirname(image_paths[index]) == current_folder:
        index += 1

    show_image()


def stop_review(b=None):
    global stopped
    stopped = True
    show_image()


# Buttons
keep_btn = Button(description="Keep (k)", button_style="success")
delete_btn = Button(description="Delete (d)", button_style="danger")
skip_btn = Button(description="Skip Folder (s)", button_style="warning")
stop_btn = Button(description="Stop (t)", button_style="info")

keep_btn.on_click(keep_image)
delete_btn.on_click(delete_image)
skip_btn.on_click(skip_folder)
stop_btn.on_click(stop_review)

# Keyboard input
keyboard = Text(placeholder="k=keep, d=delete, s=skip, t=stop")

def key_handler(change):
    key = change["new"].lower()

    if key == "k":
        keep_image()
    elif key == "d":
        delete_image()
    elif key == "s":
        skip_folder()
    elif key == "t":
        stop_review()

    keyboard.value = ""

keyboard.observe(key_handler, names="value")

ui = VBox([
    keyboard,
    out,
    HBox([keep_btn, delete_btn, skip_btn, stop_btn])
])

display(ui)

# Initial prefetch
prefetch(0)
prefetch(1)

show_image()
```
